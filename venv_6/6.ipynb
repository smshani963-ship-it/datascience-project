```python
import seaborn as sns
```

```python

df.groupby('Channel_Name')['Avg_Viewers'].mean().plot(kind='bar')
plt.title('Average Viewers by Channel')
plt.show()

df.groupby('Channel_Name')['Avg_Rating'].mean().plot(kind='bar')
plt.title('Average Rating by Channel')
plt.show()

df.groupby('Channel_Name')['Total_Likes'].mean().plot(kind='bar')
plt.title('Average Likes by Channel')
plt.show()

df.groupby('Channel_Name')['YouTube_Views'].mean().plot(kind='bar')
plt.title('Avg YouTube Views by Channel')
plt.show()

df.groupby('Show_Category')['Avg_Viewers'].mean().plot(kind='bar')
plt.title('Viewers by Show Category')
plt.show()

plt.hist(df['Avg_Viewers'], bins=20)
plt.title('Histogram of Avg_Viewers')
plt.show()

plt.hist(df['Avg_Rating'], bins=20)
plt.title('Histogram of Avg_Rating')
plt.show()

sns.boxplot(y=df['Avg_Viewers'])
plt.title('Boxplot for Avg_Viewers')
plt.show()

sns.boxplot(y=df['Total_Likes'])
plt.title('Boxplot for Total_Likes')
plt.show()

df['Channel_Category'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Channel Distribution')
plt.show()
```

```python

channel_summary = df.groupby('Channel_Name')[['Avg_Viewers','Avg_Rating','Total_Likes']].mean()
print(channel_summary)

show_category_summary = df.groupby('Show_Category')[['Avg_Viewers','Avg_Rating','Total_Likes']].mean()
print(show_category_summary)

most_viewed = df.sort_values(by='Avg_Viewers', ascending=False).head()
print(most_viewed[['Channel_Name','Popular_Show','Avg_Viewers']])

highest_rated = df.sort_values(by='Avg_Rating', ascending=False).head(10)
print(highest_rated[['Channel_Name','Popular_Show','Avg_Rating']])

social_media = df.groupby('Channel_Name')['Total_Likes'].mean().sort_values(ascending=False).head(10)
print(social_media)

youtube_reach = df.groupby('Channel_Name')['YouTube_Views'].mean().sort_values(ascending=False).head(10)
print(youtube_reach)

plt.figure(figsize=(12,6))
channel_summary.plot(kind='bar')
plt.title('Final TV Channel Performance Report')
plt.xticks(rotation=45)
plt.show()
```
