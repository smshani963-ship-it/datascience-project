```python
import pandas as pd
df=pd.read_csv("ecommerce_sales_customer_analytics_150k.csv")
print(df)
```

```python
print(df.head())
print(df.tail())
```

```python
print(df.info)
print(df.shape)
print(df.columns)
```

```python
df.isnull().sum()
```

```python
df.duplicated()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.isnull().sum()
```

```python
df['delivery_days']=df['delivery_days'].fillna(df['delivery_days'].mean)
print(df['delivery_days'])
df['estimated_delivery_days']=df['estimated_delivery_days'].fillna(df['estimated_delivery_days'].mean)
print(df['estimated_delivery_days'])
df['return_status']=df['return_status'].fillna(df['return_status'].mean)
print(df['return_status'])
```

```python
df=df[df["customer_age"]>=18]
print(df)
```

```python
print("\nmissing value after cleaning:")
print(df.isnull().sum())
```

```python
print("\nstatistical summary:")
print(df.describe())
```

```python
print("\n Customer Age:")
print(df["customer_age"].mean())

print("\n Custoer Postal Code:")
print(df["customer_postal_code"].mean())

print("\nHighest Rating:")
print(df["customer_rating"].max())

print("\nLowest Rating:")
print(df["customer_rating"].min())
```

```python
print("\n Quantity:")
print(df["quantity"].unique())
```

```python
print("\ncustomer:")
print(df["customer_age"].value_counts())
print("\ncustomer rating:")
print(df["customer_rating"].value_counts())
```

```python
df["customer_name"].value_counts().plot(kind="line")
plt.title("Customer")
plt.xlabel("customer_age")
plt.ylabel("gender")
plt.show()
```
